## 1. Python as a Calculator

Python supports all the standard arithmetic operators. The one to watch out for: exponentiation is `**`, not `^`.

| Operator | Meaning          | Example   | Result |
|----------|------------------|-----------|--------|
| `+`      | Addition         | `3 + 2`   | `5`    |
| `-`      | Subtraction      | `3 - 2`   | `1`    |
| `*`      | Multiplication   | `3 * 2`   | `6`    |
| `/`      | Division (float) | `7 / 2`   | `3.5`  |
| `//`     | Floor division   | `7 // 2`  | `3`    |
| `%`      | Modulo (remainder)| `7 % 2`  | `1`    |
| `**`     | Exponentiation   | `2 ** 10` | `1024` |

```python
# Try some arithmetic — hit Shift+Enter to run
print(3 + 2)
print(2 ** 10)          # 1024
print(17 // 3)          # floor division: 5
print(17 % 3)           # remainder: 2
print(2 ** 0.5)         # square root via fractional exponent
```

A key astronomy application: the **magnitude scale** relates the flux ratio of two stars to their magnitude difference:
$$\frac{F_1}{F_2} = 10^{\,(m_2 - m_1)\,/\,2.5}=10^{{\rm \Delta mag}/2.5}$$
Sirius ($m = -1.46$) and Vega ($m = 0.03$). How many times brighter is Sirius?

```python
m_sirius = -1.46
m_vega   =  0.03

flux_ratio = 10 ** ((m_vega - m_sirius) / 2.5)
print(flux_ratio)   # ~4.1: Sirius is about 4× brighter than Vega
```

For functions like `log10` and `sqrt`, import the `math` standard library. \
In astronomy code you'll almost always use `numpy` for this instead (covered in Section 9), but `math` is useful for quick calculations and requires no installation.

```python
import math

# Distance modulus: μ = 5 log₁₀(d / 10 pc)
d_vega_pc = 7.68
mu = 5 * math.log10(d_vega_pc / 10)
print(f"Distance modulus of Vega: {mu:.3f} mag")  # expect ≈ -0.53

# Wien's displacement law: peak wavelength of a blackbody
T_sun = 5778          # Kelvin
lambda_peak_m = 2.898e-3 / T_sun
print(f"Sun's peak wavelength: {lambda_peak_m * 1e9:.0f} nm")
```

## 2. Variables and Data Types

Assign a value to a variable with `=`. Python is **dynamically typed**, meaning you don't declare the type; Python infers it.

The four fundamental types you'll use constantly:

| Type   | Example                     | Description            |
|--------|-----------------------------|------------------------|
| `int`  | `42`, `-3`                  | Integer                |
| `float`| `3.14`, `-1.46`             | Decimal number         |
| `str`  | `"Vega"`, `'A0V'`           | Text (string)          |
| `bool` | `True`, `False`             | Boolean (logical)      |

```python
star_name      = "Vega"        # str
apparent_mag   = 0.03          # float
spectral_class = "A0V"         # str
is_variable    = False         # bool
num_planets    = 0             # int

# type() tells you the type of any object
print(type(apparent_mag))      # <class 'float'>
print(type(star_name))         # <class 'str'>

# Division always returns float in Python 3, even for whole numbers
print(10 / 2)    # 5.0   float
print(10 // 2)   # 5     int (use this when indexing arrays!)
```

## 3. Strings and F-strings

Strings are sequences of characters. You can index, slice, and call methods on them.
The modern way to embed values in strings is an **f-string**: prefix with `f` and put expressions inside `{}`. You can also add format specifiers after a `:` inside the braces.

```python
star = "Betelgeuse"

# Basic string operations
print(len(star))           # 10
print(star.upper())        # BETELGEUSE
print(star[0])             # B  (indexing — see Section 4 for more)
print(star[:5])            # Betel

# Concatenation with +
full = "α " + "Orionis"    # Betelgeuse's Bayer designation
print(full)
```

```python
# F-strings: the cleanest way to format output
T_eff  = 3500     # Kelvin
radius = 764      # solar radii

# :.0f = 0 decimal places, :.2e = scientific notation
print(f"{star} has T_eff = {T_eff} K and R = {radius} R_sun")
print(f"Luminosity ≈ {radius**2 * (T_eff/5778)**4:.2e} L_sun")
```

> **Format specifier quick reference:**
> `:.2f` → 2 decimal places &nbsp;|&nbsp; `:.3e` → scientific notation &nbsp;|&nbsp; `:8s` → string padded to 8 chars &nbsp;|&nbsp; `:+.2f` → always show sign
>
> **Raw Strings:**
> Prefix with `r` to treat backslashes as literal characters, useful for file paths and regex.
> Also very useful for LaTeX in Matplotlib, but we'll cover that in Session 3. Example: `r"\alpha"` would render as the Greek letter α in Matplotlib.

## 4. Lists, Slicing, and Tuples

### Lists

Lists are **ordered, mutable** sequences. They are the workhorse collection type in Python. \
Indexing is **0-based** (the first element is index `0`).

```python
# A mini catalog of the 5 brightest stars
star_names  = ["Sirius", "Canopus", "Arcturus", "Vega",  "Capella"]
magnitudes  = [-1.46,    -0.74,     -0.05,      0.03,    0.08    ]

print(star_names[0])    # "Sirius"  — first element
print(magnitudes[-1])   # 0.08     — last element (negative index counts from end)
print(len(star_names))  # 5
```

```python
# Modifying lists
star_names.append("Rigel")   # add to end
magnitudes.append(0.12)

print(star_names)
print(len(star_names))       # now 6
```

### Slicing

Extract a sub-list with `list[start:stop:step]`. \
The `stop` index is **exclusive** (not included). Omit `start` or `stop` to go from the beginning or to the end.

```python
print(star_names[1:4])   # index 1, 2, 3 → ["Canopus", "Arcturus", "Vega"]
print(star_names[:3])    # first 3 elements
print(star_names[-2:])   # last 2 elements
print(magnitudes[::2])   # every other element

# range() creates integer sequences — essential for loops
print(list(range(5)))          # [0, 1, 2, 3, 4]
print(list(range(1, 10, 2)))   # [1, 3, 5, 7, 9]
```

### Tuples

Tuples are like lists but **immutable**. Once they're created, they cannot be modified. \
Use them when values must not change (e.g., a coordinate pair). They also appear whenever a function returns multiple values.

```python
# (RA, Dec) for Vega — shouldn't be mutable, so a tuple fits
vega_coords = (279.23, +38.78)

ra, dec = vega_coords   # tuple unpacking — very common Python idiom
print(f"Vega: RA = {ra}°, Dec = {dec}°")

# Attempting to modify a tuple raises a TypeError:
# vega_coords[0] = 0.0   # ← uncomment to see the error
```

## 5. Dictionaries

Dictionaries store **key–value pairs** like a lookup table. \
Keys must be immutable (strings are most common); values can be anything.

```python
import math  # already imported above, but included for clarity

sirius = {
    "name":          "Sirius",
    "spectral_type": "A1V",
    "apparent_mag":  -1.46,
    "distance_pc":    2.64,
    "is_binary":      True,
}

# Access by key — raises KeyError if key doesn't exist
print(sirius["name"])
print(sirius["distance_pc"])

# Add or update a key
sirius["absolute_mag"] = sirius["apparent_mag"] - 5 * math.log10(sirius["distance_pc"] / 10)
print(f"Absolute magnitude: {sirius['absolute_mag']:.2f}")
```

```python
# Iterate over all key–value pairs
for key, value in sirius.items():
    print(f"  {key:15s}: {value}")

# Check membership
print("luminosity" in sirius)   # False
print("name" in sirius)         # True
```

## 6. Booleans and Control Flow

### Comparisons and Logical Operators

Comparison operators return `True` or `False`. Combine them with `and`, `or`, `not`.

| Operator | Meaning               |
|----------|-----------------------|
| `==`     | Equal to              |
| `!=`     | Not equal to          |
| `<`, `>` | Less / greater than   |
| `<=`,`>=`| Less / greater or equal |
| `in`     | Membership test       |

```python
m = -1.46

print(m < 0)                     # True — negative magnitude means very bright
print(m > 6.5)                   # False — 6.5 is the naked-eye limit
print(-1 < m < 0)                # True — chained comparison
print("A" in ["O","B","A","F"])  # True — membership test

# and / or / not
print(m < 0 and m > -2)   # True  (bright but not absurdly so)
print(m < -3 or m > 10)   # False (neither extreme)
```

### `if` / `elif` / `else`

```python
def classify_magnitude(m):
    """Classify a star's visibility by apparent magnitude."""
    if m < -1:
        return "Blazingly bright (e.g. Sirius, Venus)"
    elif m <= 2:
        return "Very bright naked-eye star"
    elif m <= 6.5:
        return "Visible to the naked eye"
    elif m <= 10:
        return "Binoculars needed"
    else:
        return "Telescope required"

for name, mag in zip(["Sirius", "Polaris", "Proxima Cen"],
                     [-1.46,    1.98,      11.13]):
    print(f"{name:12s} (m={mag:+.2f}): {classify_magnitude(mag)}")
```

### `for` loops

```python
star_names = ["Sirius", "Canopus", "Arcturus", "Vega", "Capella"]
magnitudes = [-1.46, -0.74, -0.05, 0.03, 0.08]

# zip() pairs two lists element-by-element
for name, mag in zip(star_names, magnitudes):
    print(f"{name:12s}  m = {mag:+.2f}")
```

```python
# enumerate() gives index + value simultaneously
for i, name in enumerate(star_names):
    print(f"  #{i+1}: {name}")
```

### `while` loops

`while` loops run as long as a condition is `True`. Prefer `for` loops when you know the number of iterations in advance. `while` loops are for when you don't.

```python
# Find the first star fainter than magnitude 0
i = 0
while magnitudes[i] < 0:
    i += 1
print(f"First star fainter than m=0: {star_names[i]} (m={magnitudes[i]})")
```

### List Comprehensions

List comprehensions build a new list by applying an expression to each element of a sequence. \
They are compact, readable, and *FASTER* than equivalent `for` + `append` loops. \
You'll see them constantly in astronomy Python code.

```python
# Distance modulus for each star
distances_pc = [2.64, 96.9, 11.26, 7.68, 42.9]

abs_mags = [m - 5 * math.log10(d / 10) for m, d in zip(magnitudes, distances_pc)]

for name, M in zip(star_names, abs_mags):
    print(f"{name:12s}  M_V = {M:+.2f}")
```

## 7. Functions

Functions package reusable logic into a named block. Two rules to live by:
1. Write a **docstring** immediately after `def`. It's how `help()` knows what your function does.
2. Follow **DRY** (Don't Repeat Yourself); if you write the same code twice, make it a function.

```python
def distance_modulus(d_pc):
    """
    Compute the distance modulus μ = m - M.

    Parameters
    ----------
    d_pc : float
        Distance in parsecs. Must be positive.

    Returns
    -------
    float
        Distance modulus in magnitudes.
    """
    return 5 * math.log10(d_pc / 10)

# Test
print(distance_modulus(10))    # 0.0 (by definition)
print(distance_modulus(100))   # 5.0
help(distance_modulus)         # prints the docstring
```

Functions can **return multiple values** as a tuple (Python unpacks them automatically).

```python
def apparent_to_absolute(m, d_pc):
    """
    Convert apparent magnitude to absolute magnitude.

    Parameters
    ----------
    m : float
        Apparent magnitude.
    d_pc : float
        Distance in parsecs.

    Returns
    -------
    M : float
        Absolute magnitude.
    mu : float
        Distance modulus.
    """
    mu = distance_modulus(d_pc)
    M  = m - mu
    return M, mu

# Unpack the two return values
M_siri, mu_siri = apparent_to_absolute(-1.46, 2.64)
print(f"Sirius: M_V = {M_siri:.2f},  μ = {mu_siri:.2f}")
```

### Default (optional) arguments

Parameters with a default value are optional, the caller only needs to provide them when \
a non-default behavior is wanted. Required parameters (no default) must come first.

```python
def wien_peak(T_kelvin, unit="nm"):
    """
    Compute the blackbody peak wavelength via Wien's displacement law.

    Parameters
    ----------
    T_kelvin : float
        Temperature in Kelvin.
    unit : str, optional
        Output unit: 'nm' (default), 'um', or 'm'.

    Returns
    -------
    float
        Peak wavelength in the requested unit.
    """
    lambda_m = 2.898e-3 / T_kelvin

    conversions = {"m": 1, "nm": 1e9, "um": 1e6}
    return lambda_m * conversions[unit]

print(f"Sun (5778 K):        {wien_peak(5778):.0f} nm  — visible light")
print(f"Betelgeuse (3500 K): {wien_peak(3500):.0f} nm  — near infrared")
print(f"Rigel (12000 K):     {wien_peak(12000):.0f} nm  — near UV")
print(f"Sun in microns:      {wien_peak(5778, unit='um'):.3f} μm")
```

The dictionary lookup `conversions[unit]` is a cleaner alternative to a long chain of `if/elif` statements. We will go over another method to handle units later on: the incredible `astropy` package.

## 8. Errors and Exceptions

When Python encounters an error it **raises an exception** and prints a traceback. \
Reading a traceback from bottom to top is one of the most important debugging skills. \
The last line tells you the error type and message; lines above it show the call stack.

Common errors you'll encounter:

| Exception | Typical cause |
|---|---|
| `TypeError` | Wrong type passed to an operation (`"m" + 3`) |
| `IndexError` | Index out of range (`my_list[99]`) |
| `KeyError` | Key not found in a dict (`d["missing_key"]`) |
| `ValueError` | Right type, wrong value (`math.log(-1)`) |
| `ZeroDivisionError` | Division by zero |
| `NameError` | Variable used before it was defined |

```python
# Deliberately trigger common exceptions — uncomment one at a time:

# TypeError
# "Vega" + 3

# IndexError
# star_names = ["Sirius", "Vega"]
# star_names[5]

# KeyError
# sirius["luminosity"]

# ValueError
# math.log(-1)
```

`try` / `except` lets you handle expected errors gracefully instead of crashing:

```python
def safe_distance(p_mas):
    """Return distance in pc, or None if the parallax is non-positive."""
    try:
        if p_mas <= 0:
            raise ValueError(f"Parallax must be positive, got {p_mas} mas")
        return 1000.0 / p_mas
    except ValueError as e:
        print(f"Warning: {e}")
        return None

print(safe_distance(768.5))    # Proxima Centauri — valid
print(safe_distance(-5.0))     # bad input — handled gracefully
```

## 9. A Taste of NumPy

NumPy is the foundation of scientific Python. Its core object is the `ndarray`, an efficient, typed array that supports mathematical operations on entire arrays at once (**broadcasting**), without needing explicit `for` loops.

You will use it in every session of this bootcamp. The standard import convention is:

```python
import numpy as np

# Create arrays from lists
mags  = np.array([-1.46, -0.74, -0.05, 0.03, 0.08, 0.12])
names = np.array(["Sirius", "Canopus", "Arcturus", "Vega", "Capella", "Rigel"])

print("dtype:", mags.dtype)    # float64
print("shape:", mags.shape)    # (6,)
```

```python
# Math on arrays applies element-wise — NO loop needed
fluxes = 10 ** ((0.03 - mags) / 2.5)   # flux relative to Vega
print("Flux ratios relative to Vega:")
print(np.round(fluxes, 2))

# Aggregation functions
print(f"Brightest: {names[np.argmin(mags)]} (m = {mags.min():.2f})")
print(f"Mean mag:  {mags.mean():.2f}")
print(f"Std dev:   {mags.std():.2f}")
```

```python
# Array creation utilities — you'll use these constantly
print(np.zeros(4))                 # [0. 0. 0. 0.]
print(np.ones(3))                  # [1. 1. 1.]
print(np.arange(0, 10, 2))         # [0 2 4 6 8]
print(np.linspace(0, 1, 5))        # [0. 0.25 0.5 0.75 1.]  — evenly spaced

# Boolean indexing — select elements that satisfy a condition
bright = mags < 0
print(names[bright])               # stars brighter than m=0
```

> **The most important NumPy habit:** replace Python `for` loops over arrays with \
> vectorised NumPy operations wherever possible. When your array has a million elements \
> (common in astronomy), this difference can mean seconds vs. hours of runtime.

## 10. A Taste of Matplotlib

Matplotlib is the standard plotting library. We'll cover it in detail in Exercise 02. \
Here is a quick preview using Planck blackbody curves. You'll see how little code it takes to produce a figure that communicates real physics.

```python
import matplotlib.pyplot as plt

def planck_flux(wavelength_nm, T_kelvin):
    """
    Planck spectral radiance (normalised to peak = 1).

    Parameters
    ----------
    wavelength_nm : ndarray
        Wavelength array in nanometres.
    T_kelvin : float
        Blackbody temperature in Kelvin.

    Returns
    -------
    ndarray
        Normalised spectral flux.
    """
    h = 6.626e-34; c = 3.0e8; k = 1.381e-23
    lam = wavelength_nm * 1e-9          # nm → m
    B   = 1 / (lam**5 * (np.exp(h*c / (lam*k*T_kelvin)) - 1))
    return B / B.max()                  # normalise

lam = np.linspace(100, 2500, 1000)     # wavelength axis (nm)

# Stars of different spectral types
stars = {
    "O-star  (30 000 K)":  30000,
    "G-star Sun (5 778 K)": 5778,
    "M-star  (3 500 K)":    3500,
}

fig, ax = plt.subplots(figsize=(8, 4))

for label, T in stars.items():
    ax.plot(lam, planck_flux(lam, T), label=label)

# Shade the visible range
ax.axvspan(380, 700, alpha=0.1, color="gold", label="Visible (380–700 nm)")

ax.set_xlabel("Wavelength (nm)")
ax.set_ylabel("Relative Flux")
ax.set_title("Blackbody Spectra for Different Stellar Types")
ax.legend()
plt.tight_layout()
plt.show()
```
